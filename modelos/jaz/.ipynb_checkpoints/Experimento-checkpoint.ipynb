```python
#datos
from ortools.sat.python import cp_model
import networkx as nx
import matplotlib.pyplot as plt
from itertools import combinations

# ============================================================
# 1. NODOS
# ============================================================

B = "B"   # Base
D = "D"   # Botadero

urban_nodes = [str(i) for i in range(1, 11)]
V = [B] + urban_nodes + [D]

# ============================================================
# 2. ARCOS DE LA RED A
# ============================================================

A = [
    ("B","1"), ("1","B"),
    ("B","6"), ("6","B"),

    ("1","2"), ("2","1"),
    ("1","6"), ("6","1"),

    ("2","3"), ("3","2"),

    ("3","6"),
    ("3","4"),
    ("4","3"),       # agregado para garantizar conectividad dirigida

    ("6","7"),
    ("7","4"),

    ("4","5"), ("5","4"),

    ("5","7"), ("7","5"),

    ("4","8"),

    ("8","9"), ("9","8"),

    ("8","10"),
    ("10","9"),

    ("5","10"), ("10","5"),

    ("9","D"), ("D","9"),
    ("10","D"), ("D","10")
]

# Eliminar posibles duplicados conservando orden
A = list(dict.fromkeys(A))

# ============================================================
# 3. ARCOS QUE REQUIEREN SERVICIO A_R
# ============================================================

AR = [
    ("1","2"),
    ("2","1"),
    ("2","3"),
    ("3","2"),
    ("3","6"),
    ("6","7"),
    ("7","4"),
    ("4","5"),
    ("5","7"),
    ("4","8"),
    ("8","10"),
    ("10","9")
]

# ============================================================
# 4. DEMANDAS q_ij
# ============================================================

q = {
    ("1","2"): 5,
    ("2","1"): 3,
    ("2","3"): 4,
    ("3","2"): 2,
    ("3","6"): 4,
    ("6","7"): 3,
    ("7","4"): 5,
    ("4","5"): 4,
    ("5","7"): 2,
    ("4","8"): 3,
    ("8","10"): 4,
    ("10","9"): 3
}

# ============================================================
# 5. COSTOS DE DEADHEADING d_ij
# ============================================================

d = {
    ("B","1"): 2, ("1","B"): 2,
    ("B","6"): 2, ("6","B"): 2,

    ("1","2"): 2, ("2","1"): 2,
    ("1","6"): 3, ("6","1"): 3,

    ("2","3"): 2, ("3","2"): 2,

    ("3","6"): 2,
    ("3","4"): 3,
    ("4","3"): 3,

    ("6","7"): 2,
    ("7","4"): 2,

    ("4","5"): 2, ("5","4"): 2,

    ("5","7"): 2, ("7","5"): 2,

    ("4","8"): 3,

    ("8","9"): 2, ("9","8"): 2,

    ("8","10"): 2,
    ("10","9"): 2,

    ("5","10"): 3, ("10","5"): 3,

    ("9","D"): 4, ("D","9"): 4,
    ("10","D"): 3, ("D","10"): 3
}

# ============================================================
# 6. COSTOS DE SERVICIO c_ij
# ============================================================

c = {
    ("1","2"): 3,
    ("2","1"): 3,
    ("2","3"): 3,
    ("3","2"): 3,
    ("3","6"): 3,
    ("6","7"): 3,
    ("7","4"): 3,
    ("4","5"): 3,
    ("5","7"): 3,
    ("4","8"): 4,
    ("8","10"): 3,
    ("10","9"): 3
}

# ============================================================
# 7. PARÁMETROS
# ============================================================

P = 2
W = 10
L = 6

R = range(1, L + 1)

initial_trips = [1, 2]
final_trips = [3, 4]
intermediate_trips = [5, 6]

QT = sum(q[a] for a in AR)

print("Número de nodos:", len(V))
print("Número de arcos:", len(A))
print("Número de arcos requeridos:", len(AR))
print("Demanda total QT =", QT)
print("Capacidad W =", W)
print("Vehículos P =", P)
print("Trips potenciales L =", L)
print("Cota inferior por capacidad =", (QT + W - 1)//W)
```

```python
assert set(AR).issubset(set(A))
assert set(q.keys()) == set(AR)
assert set(c.keys()) == set(AR)
assert set(d.keys()) == set(A)

for a in AR:
    assert q[a] > 0
    assert d[a] < c[a], f"No se cumple d < c en {a}"

print("✓ Datos consistentes.")
```

```python
#crear variables
model = cp_model.CpModel()

# x[r,i,j] = 1 si el arco requerido (i,j)
# se atiende en el trip r
x = {}

for r in R:
    for i, j in AR:
        x[r,i,j] = model.NewBoolVar(f"x_{r}_{i}_{j}")

# y[r,i,j] = número de veces que se atraviesa (i,j)
# sin recoger basura en trip r.
#
# Para esta instancia pequeña usamos una cota superior suficientemente
# holgada. Una solución óptima nunca debería necesitar recorrer un mismo
# arco una cantidad enorme de veces.
YMAX = len(A)

y = {}

for r in R:
    for i, j in A:
        y[r,i,j] = model.NewIntVar(0, YMAX, f"y_{r}_{i}_{j}")

print("Variables x:", len(x))
print("Variables y:", len(y))
```

```python
#funcion objervo
service_cost = sum(
    c[i,j] * x[r,i,j]
    for r in R
    for i,j in AR
)

deadhead_cost = sum(
    d[i,j] * y[r,i,j]
    for r in R
    for i,j in A
)

model.Minimize(service_cost + deadhead_cost)
```

```python
#función auxiliar para flujo
#Esto nos facilitará escribir las restricciones.
def service_in(r, node):
    return sum(
        x[r,i,j]
        for i,j in AR
        if j == node
    )

def service_out(r, node):
    return sum(
        x[r,i,j]
        for i,j in AR
        if i == node
    )

def dead_in(r, node):
    return sum(
        y[r,i,j]
        for i,j in A
        if j == node
    )

def dead_out(r, node):
    return sum(
        y[r,i,j]
        for i,j in A
        if i == node
    )
```

```python
#restricción (2): continuidad
for r in R:
    for node in urban_nodes:

        model.Add(
            service_in(r,node)
            + dead_in(r,node)
            - service_out(r,node)
            - dead_out(r,node)
            == 0
        )
```

```python
#restricciones (3)–(11)
#Ahora codificamos initial, final e intermediate trips
# ============================================================
# INITIAL TRIPS r = 1,2
# B -> ... -> D
# ============================================================

for r in initial_trips:

    # (3) Sale netamente una unidad de B
    model.Add(
        dead_out(r,B) - dead_in(r,B) == 1
    )

    # (4) entra exactamente una vez a D
    model.Add(
        dead_in(r,D) == 1
    )

    # (5) no sale de D
    model.Add(
        dead_out(r,D) == 0
    )


# ============================================================
# FINAL TRIPS r = 3,4
# D -> ... -> B
# ============================================================

for r in final_trips:

    # (6) entra netamente una unidad a B
    model.Add(
        dead_in(r,B) - dead_out(r,B) == 1
    )

    # (7) sale exactamente una vez de D
    model.Add(
        dead_out(r,D) == 1
    )

    # (8) no vuelve a entrar a D
    model.Add(
        dead_in(r,D) == 0
    )


# ============================================================
# INTERMEDIATE TRIPS r = 5,6
# D -> ... -> D, o trip no utilizado
# ============================================================

for r in intermediate_trips:

    # (9) balance cero en B
    model.Add(
        dead_out(r,B) - dead_in(r,B) == 0
    )

    # (10) balance cero en D
    model.Add(
        dead_out(r,D) - dead_in(r,D) == 0
    )

    # (11) sale como máximo una vez de D
    model.Add(
        dead_out(r,D) <= 1
    )
```

```python
#restricciones (3)–(11)
#Ahora codificamos initial, final e intermediate trips.
# ============================================================
# INITIAL TRIPS r = 1,2
# B -> ... -> D
# ============================================================

for r in initial_trips:

    # (3) Sale netamente una unidad de B
    model.Add(
        dead_out(r,B) - dead_in(r,B) == 1
    )

    # (4) entra exactamente una vez a D
    model.Add(
        dead_in(r,D) == 1
    )

    # (5) no sale de D
    model.Add(
        dead_out(r,D) == 0
    )


# ============================================================
# FINAL TRIPS r = 3,4
# D -> ... -> B
# ============================================================

for r in final_trips:

    # (6) entra netamente una unidad a B
    model.Add(
        dead_in(r,B) - dead_out(r,B) == 1
    )

    # (7) sale exactamente una vez de D
    model.Add(
        dead_out(r,D) == 1
    )

    # (8) no vuelve a entrar a D
    model.Add(
        dead_in(r,D) == 0
    )


# ============================================================
# INTERMEDIATE TRIPS r = 5,6
# D -> ... -> D, o trip no utilizado
# ============================================================

for r in intermediate_trips:

    # (9) balance cero en B
    model.Add(
        dead_out(r,B) - dead_in(r,B) == 0
    )

    # (10) balance cero en D
    model.Add(
        dead_out(r,D) - dead_in(r,D) == 0
    )

    # (11) sale como máximo una vez de D
    model.Add(
        dead_out(r,D) <= 1
    )
```

```python
#rrestricción (14): cada arco requerido exactamente una vez
for i,j in AR:

    model.Add(
        sum(
            x[r,i,j]
            for r in R
        )
        == 1
    )
```

```python
#conectividad, equivalente computacional de (12)
#Aquí debemos tener especial cuidado. Vamos a generar las restricciones para todos los subconjuntos no vacíos de nodos urbanos. Para 10 nodos son:
#\[
#2^{10}-1=1023
#\] subconjuntos por trip, totalmente manejable.
# ============================================================
# RESTRICCIONES DE CONECTIVIDAD
# Basadas en la lógica de la restricción (12)
# ============================================================

M = len(A) * YMAX + len(AR)

count_connectivity = 0

for r in R:

    for size in range(1, len(urban_nodes) + 1):

        for subset_tuple in combinations(urban_nodes, size):

            S = set(subset_tuple)

            # Arcos utilizados completamente dentro de S
            internal_service = sum(
                x[r,i,j]
                for i,j in AR
                if i in S and j in S
            )

            internal_dead = sum(
                y[r,i,j]
                for i,j in A
                if i in S and j in S
            )

            # Arcos utilizados que cruzan la frontera de S
            crossing_service = sum(
                x[r,i,j]
                for i,j in AR
                if (i in S) != (j in S)
            )

            crossing_dead = sum(
                y[r,i,j]
                for i,j in A
                if (i in S) != (j in S)
            )

            model.Add(
                internal_service + internal_dead
                <=
                M * (crossing_service + crossing_dead)
            )

            count_connectivity += 1

print("Restricciones de conectividad:", count_connectivity)
```

Para nuestro experimento pequeño esto nos permite impedir componentes aisladas. Conceptualmente, (12) del artículo tiene justamente la finalidad de impedir viajes ilegales que no contienen el nodo de descarga.

```python
solver = cp_model.CpSolver()

# Tiempo máximo
solver.parameters.max_time_in_seconds = 120

# Usar varios núcleos
solver.parameters.num_search_workers = 8

# Mostrar progreso si quieres
solver.parameters.log_search_progress = False

status = solver.Solve(model)

print("=" * 60)

if status == cp_model.OPTIMAL:
    print("✓ SOLUCIÓN ÓPTIMA")
elif status == cp_model.FEASIBLE:
    print("✓ SOLUCIÓN FACTIBLE (no se probó optimalidad)")
elif status == cp_model.INFEASIBLE:
    print("✗ MODELO INFACTIBLE")
else:
    print("Estado:", solver.StatusName(status))

if status in [cp_model.OPTIMAL, cp_model.FEASIBLE]:

    print(f"\nCosto total = {solver.ObjectiveValue():.0f}")

    sc = sum(
        c[i,j] * solver.Value(x[r,i,j])
        for r in R
        for i,j in AR
    )

    dc = sum(
        d[i,j] * solver.Value(y[r,i,j])
        for r in R
        for i,j in A
    )

    print("Costo servicio    =", sc)
    print("Costo deadheading =", dc)
```

```python
#examinar qué hace cada trip
#Esta es la celda más importante para entender el resultado.
if status in [cp_model.OPTIMAL, cp_model.FEASIBLE]:

    for r in R:

        print("\n" + "="*65)

        if r in initial_trips:
            tipo = "INICIAL"
        elif r in final_trips:
            tipo = "FINAL"
        else:
            tipo = "INTERMEDIO"

        print(f"TRIP {r} | {tipo}")
        print("="*65)

        serviced = []
        deadheaded = []

        total_q = 0
        trip_service_cost = 0
        trip_dead_cost = 0

        for i,j in AR:

            if solver.Value(x[r,i,j]) == 1:

                serviced.append((i,j))

                total_q += q[i,j]
                trip_service_cost += c[i,j]

        for i,j in A:

            repetitions = solver.Value(y[r,i,j])

            if repetitions > 0:

                deadheaded.append((i,j,repetitions))
                trip_dead_cost += d[i,j] * repetitions

        print("\nArcos SERVIDOS:")

        if serviced:

            for i,j in serviced:

                print(
                    f"  {i:>2} -> {j:<2}"
                    f" | demanda = {q[i,j]}"
                    f" | costo = {c[i,j]}"
                )

        else:
            print("  Ninguno")

        print("\nArcos DEADHEADING:")

        if deadheaded:

            for i,j,k in deadheaded:

                print(
                    f"  {i:>2} -> {j:<2}"
                    f" | veces = {k}"
                    f" | costo unitario = {d[i,j]}"
                )

        else:
            print("  Ninguno")

        print("\nResumen:")
        print(f"  Demanda recogida = {total_q}/{W}")
        print(f"  Costo servicio   = {trip_service_cost}")
        print(f"  Costo deadhead   = {trip_dead_cost}")
        print(
            f"  Costo trip       = "
            f"{trip_service_cost + trip_dead_cost}"
        )
```

```python
#comprobar que no perdimos basura
print("\nVERIFICACIÓN DE ARCOS REQUERIDOS")
print("="*70)

for i,j in AR:

    trips = [
        r
        for r in R
        if solver.Value(x[r,i,j]) == 1
    ]

    print(
        f"{i:>2} -> {j:<2}"
        f" | q={q[i,j]}"
        f" | atendido en trip {trips}"
    )

print("\nDemanda total atendida:")

total_served = sum(
    q[i,j] * solver.Value(x[r,i,j])
    for r in R
    for i,j in AR
)

print(total_served, "/", QT)
```

```python
G = nx.DiGraph()

G.add_nodes_from(V)
G.add_edges_from(A)

pos = {
    "B":  (-1, 0),
    "1":  (0, 0),
    "2":  (0, 2),
    "3":  (2, 2),
    "4":  (4, 2),
    "5":  (6, 2),
    "6":  (2, 0),
    "7":  (4, 0),
    "8":  (4, 4),
    "9":  (6, 4),
    "10": (6, 0),
    "D":  (8, 4)
}

plt.figure(figsize=(14, 9))

nx.draw_networkx_nodes(
    G,
    pos,
    node_size=1200
)

nx.draw_networkx_labels(
    G,
    pos,
    font_size=11,
    font_weight="bold"
)

# Arcos no requeridos
non_required = [a for a in A if a not in AR]

nx.draw_networkx_edges(
    G,
    pos,
    edgelist=non_required,
    arrows=True,
    arrowsize=18,
    width=1.5,
    connectionstyle="arc3,rad=0.08"
)

# Arcos requeridos
nx.draw_networkx_edges(
    G,
    pos,
    edgelist=AR,
    arrows=True,
    arrowsize=22,
    width=3,
    connectionstyle="arc3,rad=0.08"
)

# Etiquetas q sobre arcos requeridos
edge_labels = {
    (i,j): f"q={q[i,j]}"
    for i,j in AR
}

nx.draw_networkx_edge_labels(
    G,
    pos,
    edge_labels=edge_labels,
    font_size=9,
    rotate=False
)

plt.title(
    "Instancia CARP experimental\n"
    "Arcos gruesos = calles que requieren recolección"
)

plt.axis("off")
plt.show()
```
